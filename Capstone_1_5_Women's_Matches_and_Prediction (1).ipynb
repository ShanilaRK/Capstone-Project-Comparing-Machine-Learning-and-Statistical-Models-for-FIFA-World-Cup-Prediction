```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)

DATA_DIR = "sample_data"

print("=" * 70)
print("NOTEBOOK 5: WOMEN'S WORLD CUP + EURO DATA PREPARATION")
print("=" * 70)
```

```python
import os
import glob

DATA_DIR = "/content/sample_data"

# find CSV files named by year
all_csv_files = glob.glob(
    os.path.join(DATA_DIR, "*.csv")
)

euro_files = []

for file in all_csv_files:

    filename = os.path.basename(file)

    # only keep files whose name is a 4-digit year
    year_text = filename.replace(".csv", "")

    if year_text.isdigit() and len(year_text) == 4:
        euro_files.append(file)

# chronologically
euro_files = sorted(
    euro_files,
    key=lambda x: int(
        os.path.basename(x).replace(".csv", "")
    )
)

print("=" * 60)
print("WOMEN'S EURO FILES FOUND")
print("=" * 60)

print("Number of files:", len(euro_files))

for file in euro_files:
    print("✓", os.path.basename(file))
```

```python
euro_dfs = []

for file in euro_files:

    filename = os.path.basename(file)

    year = int(
        filename.replace(".csv", "")
    )

    df = pd.read_csv(file)

    # tournament info
    df["tournament"] = "Women's EURO"
    if "Year" not in df.columns:
        df["Year"] = year

    euro_dfs.append(df)

    print(
        f"✓ Loaded {filename}: "
        f"{len(df)} matches"
    )


womens_euro = pd.concat(
    euro_dfs,
    ignore_index=True
)

print("\n" + "=" * 60)
print("EURO DATA LOADED")
print("=" * 60)

print(
    "Total matches:",
    len(womens_euro)
)

print(
    "Shape:",
    womens_euro.shape
)

display(
    womens_euro.head()
)
```

```python
def clean_columns(df):

    df = df.copy()

    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("-", "_")
        .str.replace(".", "_", regex=False)
    )

    return df


womens_euro = clean_columns(
    womens_euro
)

womens_world_cup = clean_columns(
    womens_world_cup
)

print("EURO columns:")
print(
    womens_euro.columns.tolist()
)

print("\nWorld Cup columns:")
print(
    womens_world_cup.columns.tolist()
)
```

```python
CORE_COLUMNS = [
    "home_team",
    "away_team",

    "home_score",
    "away_score",

    "home_xg",
    "away_xg",

    "home_penalty",
    "away_penalty",

    "home_manager",
    "away_manager",

    "home_captain",
    "away_captain",

    "attendance",
    "venue",
    "round",
    "date",
    "score",
    "referee",
    "notes",
    "host",
    "year",
    "tournament"
]


def ensure_columns(df, columns):

    df = df.copy()

    for col in columns:

        if col not in df.columns:
            df[col] = np.nan

    return df


womens_euro = ensure_columns(
    womens_euro,
    CORE_COLUMNS
)

womens_world_cup = ensure_columns(
    womens_world_cup,
    CORE_COLUMNS
)
```

```python
womens_euro = womens_euro.loc[
    :, ~womens_euro.columns.duplicated()
].copy()

womens_world_cup = womens_world_cup.loc[
    :, ~womens_world_cup.columns.duplicated()
].copy()


print("Duplicate columns removed.")

print("\nEURO duplicate columns remaining:")
print(womens_euro.columns[womens_euro.columns.duplicated()].tolist())

print("\nWORLD CUP duplicate columns remaining:")
print(
    womens_world_cup.columns[
        womens_world_cup.columns.duplicated()
    ].tolist()
)


for col in CORE_COLUMNS:

    if col not in womens_euro.columns:
        womens_euro[col] = np.nan

    if col not in womens_world_cup.columns:
        womens_world_cup[col] = np.nan


womens_matches = pd.concat(
    [
        womens_world_cup[CORE_COLUMNS],
        womens_euro[CORE_COLUMNS]
    ],
    ignore_index=True
)


print("\n" + "=" * 60)
print("COMBINED WOMEN'S DATASET")
print("=" * 60)

print("Shape:", womens_matches.shape)

print("\nMatches by tournament:")
print(
    womens_matches["tournament"]
    .value_counts()
)

display(
    womens_matches.head(10)
)
```

```python
def knockout_flag(round_value):

    text = str(
        round_value
    ).lower()

    knockout_words = [
        "final",
        "semi",
        "quarter",
        "round of 16",
        "round of 32",
        "third",
        "playoff"
    ]

    return int(
        any(
            word in text
            for word in knockout_words
        )
    )


womens_matches["is_knockout"] = (
    womens_matches[
        "round"
    ].apply(
        knockout_flag
    )
)

print(
    womens_matches[
        "is_knockout"
    ].value_counts()
)
```

```python
from collections import defaultdict, deque
from scipy.stats import poisson

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression, PoissonRegressor
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    log_loss,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

try:
    from xgboost import XGBClassifier
    XGBOOST_AVAILABLE = True

except ImportError:
    XGBOOST_AVAILABLE = False
    print("Install XGBoost using: !pip install -q xgboost")


RANDOM_STATE = 42

CLASS_NAMES = [
    "Team B Win",
    "Draw",
    "Team A Win"
]

print("Modeling libraries loaded successfully.")
print("XGBoost available:", XGBOOST_AVAILABLE)
```

```python
def build_prematch_features(
    matches,
    window=5,
    initial_elo=1500.0,
    k_factor=24.0
):

    data = (
        matches
        .sort_values(["date", "year"])
        .reset_index(drop=True)
        .copy()
    )

    # each team begins with the same Elo rating
    elo = defaultdict(
        lambda: initial_elo
    )

    form_points = defaultdict(
        lambda: deque(maxlen=window)
    )

    goals_for = defaultdict(
        lambda: deque(maxlen=window)
    )

    goals_against = defaultdict(
        lambda: deque(maxlen=window)
    )

    feature_rows = []

    def recent_average(values, default=1.0):

        if len(values) == 0:
            return float(default)

        return float(
            np.mean(values)
        )


    for _, match in data.iterrows():

        team_a = match["team_a"]
        team_b = match["team_b"]

        score_a = float(
            match["score_a"]
        )

        score_b = float(
            match["score_b"]
        )

        points_a = list(
            form_points[team_a]
        )

        points_b = list(
            form_points[team_b]
        )

        goals_for_a = list(
            goals_for[team_a]
        )

        goals_for_b = list(
            goals_for[team_b]
        )

        goals_against_a = list(
            goals_against[team_a]
        )

        goals_against_b = list(
            goals_against[team_b]
        )

        row = match.to_dict()

        row["elo_a"] = elo[team_a]
        row["elo_b"] = elo[team_b]

        row["elo_diff"] = (
            elo[team_a]
            -
            elo[team_b]
        )

        row["form_points_a"] = recent_average(
            points_a
        )

        row["form_points_b"] = recent_average(
            points_b
        )

        row["form_diff"] = (
            row["form_points_a"]
            -
            row["form_points_b"]
        )

        row["avg_goals_for_a"] = recent_average(
            goals_for_a
        )

        row["avg_goals_for_b"] = recent_average(
            goals_for_b
        )

        row["goals_for_diff"] = (
            row["avg_goals_for_a"]
            -
            row["avg_goals_for_b"]
        )

        row["avg_goals_against_a"] = recent_average(
            goals_against_a
        )

        row["avg_goals_against_b"] = recent_average(
            goals_against_b
        )

        row["goals_against_diff"] = (
            row["avg_goals_against_a"]
            -
            row["avg_goals_against_b"]
        )

        row["win_rate_a"] = (
            np.mean(
                np.array(points_a) == 3
            )
            if points_a
            else 0.33
        )

        row["win_rate_b"] = (
            np.mean(
                np.array(points_b) == 3
            )
            if points_b
            else 0.33
        )

        row["win_rate_diff"] = (
            row["win_rate_a"]
            -
            row["win_rate_b"]
        )

        row["recent_games_a"] = len(
            points_a
        )

        row["recent_games_b"] = len(
            points_b
        )

        feature_rows.append(row)


        if score_a > score_b:

            actual_a = 1.0
            match_points_a = 3
            match_points_b = 0

        elif score_a < score_b:

            actual_a = 0.0
            match_points_a = 0
            match_points_b = 3

        else:

            actual_a = 0.5
            match_points_a = 1
            match_points_b = 1


        expected_a = (
            1
            /
            (
                1
                +
                10 ** (
                    (
                        elo[team_b]
                        -
                        elo[team_a]
                    )
                    / 400
                )
            )
        )

        elo_change = (
            k_factor
            *
            (
                actual_a
                -
                expected_a
            )
        )

        elo[team_a] += elo_change
        elo[team_b] -= elo_change


        form_points[team_a].append(
            match_points_a
        )

        form_points[team_b].append(
            match_points_b
        )

        goals_for[team_a].append(
            score_a
        )

        goals_for[team_b].append(
            score_b
        )

        goals_against[team_a].append(
            score_b
        )

        goals_against[team_b].append(
            score_a
        )


    return pd.DataFrame(
        feature_rows
    )


model_data = build_prematch_features(
    womens_matches
)

print(
    "Modeling dataset shape:",
    model_data.shape
)

display(
    model_data[
        [
            "date",
            "team_a",
            "team_b",
            "elo_a",
            "elo_b",
            "elo_diff",
            "form_diff",
            "goals_for_diff",
            "win_rate_diff",
            "target"
        ]
    ].head(10)
)
```

```python
MODEL_FEATURES = [
    "elo_diff",
    "form_diff",
    "goals_for_diff",
    "goals_against_diff",
    "win_rate_diff",
    "avg_goals_for_a",
    "avg_goals_for_b",
    "avg_goals_against_a",
    "avg_goals_against_b",
    "recent_games_a",
    "recent_games_b",
    "is_knockout"
]


# train using tournaments through 2019
train_mask = (
    model_data["year"]
    <= 2019
)

# test using 2021, 2023, and 2025 tournaments
test_mask = (
    model_data["year"]
    >= 2021
)


train_data = (
    model_data.loc[
        train_mask
    ]
    .copy()
)

test_data = (
    model_data.loc[
        test_mask
    ]
    .copy()
)


X_train = train_data[
    MODEL_FEATURES
]

y_train = (
    train_data["target"]
    .astype(int)
)

X_test = test_data[
    MODEL_FEATURES
]

y_test = (
    test_data["target"]
    .astype(int)
)


print("=" * 60)
print("CHRONOLOGICAL TRAIN/TEST SPLIT")
print("=" * 60)

print(
    "Training matches:",
    len(train_data)
)

print(
    "Training years:",
    train_data["year"].min(),
    "to",
    train_data["year"].max()
)

print()

print(
    "Testing matches:",
    len(test_data)
)

print(
    "Testing years:",
    test_data["year"].min(),
    "to",
    test_data["year"].max()
)

print("\nTraining target distribution:")

print(
    y_train
    .value_counts()
    .sort_index()
)

print("\nTesting target distribution:")

print(
    y_test
    .value_counts()
    .sort_index()
)


# Safety checks
assert len(train_data) > 0, (
    "Training dataset is empty."
)

assert len(test_data) > 0, (
    "Testing dataset is empty."
)

assert y_train.nunique() == 3, (
    "Training data must contain all three result classes."
)

assert not X_train.columns.duplicated().any(), (
    "Duplicate model feature columns detected."
)

print("\n✓ Train/test split completed successfully.")
```

```python
# reset stored results before rerunning the models
model_results = []

model_predictions = {}

trained_models = {}


def evaluate_classifier(
    model_name,
    model,
    X_train,
    y_train,
    X_test,
    y_test
):

    # train model
    model.fit(
        X_train,
        y_train
    )

    # predict result classes
    predictions = model.predict(
        X_test
    )

    # obtain predicted probabilities
    raw_probabilities = model.predict_proba(
        X_test
    )

    # force probabilities into float64
    raw_probabilities = np.asarray(
        raw_probabilities,
        dtype=np.float64
    )

    probabilities = np.zeros(
        (
            len(X_test),
            3
        ),
        dtype=np.float64
    )


    for column_position, class_label in enumerate(
        model.classes_
    ):

        probabilities[
            :,
            int(class_label)
        ] = raw_probabilities[
            :,
            column_position
        ]



    probabilities = np.clip(
        probabilities,
        1e-15,
        1.0
    )



    probabilities = (
        probabilities
        /
        probabilities.sum(
            axis=1,
            keepdims=True
        )
    )


    accuracy = accuracy_score(
        y_test,
        predictions
    )

    macro_f1 = f1_score(
        y_test,
        predictions,
        average="macro"
    )

    probability_log_loss = log_loss(
        y_test,
        probabilities,
        labels=[
            0,
            1,
            2
        ]
    )


    model_results.append(
        {
            "Model": model_name,
            "Accuracy": accuracy,
            "Macro F1": macro_f1,
            "Log Loss": probability_log_loss
        }
    )


    model_predictions[
        model_name
    ] = {
        "predictions": predictions,
        "probabilities": probabilities
    }


    trained_models[
        model_name
    ] = model


    print("=" * 60)
    print(model_name)
    print("=" * 60)

    print(
        f"Accuracy: {accuracy:.4f}"
    )

    print(
        f"Macro F1: {macro_f1:.4f}"
    )

    print(
        f"Log Loss: {probability_log_loss:.4f}"
    )

    print()

    print(
        classification_report(
            y_test,
            predictions,
            labels=[
                0,
                1,
                2
            ],
            target_names=CLASS_NAMES,
            zero_division=0
        )
    )


    probability_sums = probabilities.sum(
        axis=1
    )

    print(
        "Probability rows correctly normalized:",
        np.allclose(
            probability_sums,
            1.0
        )
    )


    return model


print(
    "✓ Corrected evaluation function created."
)
```

```python
logistic_model = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=3000,
                class_weight="balanced",
                random_state=RANDOM_STATE
            )
        )
    ]
)


logistic_model = evaluate_classifier(
    model_name="Logistic Regression",
    model=logistic_model,
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test
)
```

```python
POISSON_FEATURES = [
    "elo_diff",
    "form_diff",
    "avg_goals_for_a",
    "avg_goals_for_b",
    "avg_goals_against_a",
    "avg_goals_against_b",
    "is_knockout"
]


# separate models predict each team's expected goals
poisson_team_a = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "regressor",
            PoissonRegressor(
                alpha=0.5,
                max_iter=1000
            )
        )
    ]
)


poisson_team_b = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),
        (
            "scaler",
            StandardScaler()
        ),
        (
            "regressor",
            PoissonRegressor(
                alpha=0.5,
                max_iter=1000
            )
        )
    ]
)


# train one model for each team's goals
poisson_team_a.fit(
    train_data[
        POISSON_FEATURES
    ],
    train_data[
        "score_a"
    ]
)


poisson_team_b.fit(
    train_data[
        POISSON_FEATURES
    ],
    train_data[
        "score_b"
    ]
)


# predict expected goals
expected_goals_a = poisson_team_a.predict(
    test_data[
        POISSON_FEATURES
    ]
)

expected_goals_b = poisson_team_b.predict(
    test_data[
        POISSON_FEATURES
    ]
)


# prevent unrealistic or invalid expected-goal values
expected_goals_a = np.clip(
    expected_goals_a,
    0.05,
    8.0
)

expected_goals_b = np.clip(
    expected_goals_b,
    0.05,
    8.0
)


def calculate_poisson_result_probabilities(
    lambda_a,
    lambda_b,
    maximum_goals=10
):

    possible_goals = np.arange(
        maximum_goals + 1
    )

    probability_a = poisson.pmf(
        possible_goals,
        lambda_a
    )

    probability_b = poisson.pmf(
        possible_goals,
        lambda_b
    )

    score_probability_matrix = np.outer(
        probability_a,
        probability_b
    )

    # rows are Team A goals and columns are Team B goals
    team_a_win_probability = np.tril(
        score_probability_matrix,
        k=-1
    ).sum()

    draw_probability = np.trace(
        score_probability_matrix
    )

    team_b_win_probability = np.triu(
        score_probability_matrix,
        k=1
    ).sum()


    result_probabilities = np.array(
        [
            team_b_win_probability,
            draw_probability,
            team_a_win_probability
        ],
        dtype=np.float64
    )


    # normalize probabilities
    result_probabilities = (
        result_probabilities
        /
        result_probabilities.sum()
    )

    return result_probabilities


poisson_probabilities = np.vstack(
    [
        calculate_poisson_result_probabilities(
            lambda_a,
            lambda_b
        )
        for lambda_a, lambda_b in zip(
            expected_goals_a,
            expected_goals_b
        )
    ]
)


poisson_predictions = np.argmax(
    poisson_probabilities,
    axis=1
)


poisson_accuracy = accuracy_score(
    y_test,
    poisson_predictions
)

poisson_macro_f1 = f1_score(
    y_test,
    poisson_predictions,
    average="macro"
)

poisson_log_loss = log_loss(
    y_test,
    poisson_probabilities,
    labels=[
        0,
        1,
        2
    ]
)


model_results.append(
    {
        "Model": "Poisson",
        "Accuracy": poisson_accuracy,
        "Macro F1": poisson_macro_f1,
        "Log Loss": poisson_log_loss
    }
)


model_predictions[
    "Poisson"
] = {
    "predictions": poisson_predictions,
    "probabilities": poisson_probabilities
}


trained_models[
    "Poisson Team A Goals"
] = poisson_team_a

trained_models[
    "Poisson Team B Goals"
] = poisson_team_b


print("=" * 60)
print("Poisson")
print("=" * 60)

print(
    f"Accuracy: {poisson_accuracy:.4f}"
)

print(
    f"Macro F1: {poisson_macro_f1:.4f}"
)

print(
    f"Log Loss: {poisson_log_loss:.4f}"
)

print()

print(
    classification_report(
        y_test,
        poisson_predictions,
        labels=[
            0,
            1,
            2
        ],
        target_names=CLASS_NAMES,
        zero_division=0
    )
)

print(
    "Average predicted Team A goals:",
    round(
        expected_goals_a.mean(),
        3
    )
)

print(
    "Average predicted Team B goals:",
    round(
        expected_goals_b.mean(),
        3
    )
)
```

```python
def calculate_current_team_profiles(
    matches,
    window=5,
    initial_elo=1500.0,
    k_factor=24.0
):

    matches = (
        matches
        .sort_values(
            [
                "date",
                "year"
            ]
        )
        .reset_index(
            drop=True
        )
    )


    elo = defaultdict(
        lambda: initial_elo
    )

    form_points = defaultdict(
        lambda: deque(
            maxlen=window
        )
    )

    goals_for = defaultdict(
        lambda: deque(
            maxlen=window
        )
    )

    goals_against = defaultdict(
        lambda: deque(
            maxlen=window
        )
    )

    matches_played = defaultdict(
        int
    )


    for _, match in matches.iterrows():

        team_a = match[
            "team_a"
        ]

        team_b = match[
            "team_b"
        ]

        score_a = float(
            match[
                "score_a"
            ]
        )

        score_b = float(
            match[
                "score_b"
            ]
        )


        if score_a > score_b:

            actual_a = 1.0
            points_a = 3
            points_b = 0

        elif score_a < score_b:

            actual_a = 0.0
            points_a = 0
            points_b = 3

        else:

            actual_a = 0.5
            points_a = 1
            points_b = 1


        expected_a = (
            1
            /
            (
                1
                +
                10 ** (
                    (
                        elo[
                            team_b
                        ]
                        -
                        elo[
                            team_a
                        ]
                    )
                    / 400
                )
            )
        )


        elo_change = (
            k_factor
            *
            (
                actual_a
                -
                expected_a
            )
        )


        elo[
            team_a
        ] += elo_change

        elo[
            team_b
        ] -= elo_change


        form_points[
            team_a
        ].append(
            points_a
        )

        form_points[
            team_b
        ].append(
            points_b
        )


        goals_for[
            team_a
        ].append(
            score_a
        )

        goals_for[
            team_b
        ].append(
            score_b
        )


        goals_against[
            team_a
        ].append(
            score_b
        )

        goals_against[
            team_b
        ].append(
            score_a
        )


        matches_played[
            team_a
        ] += 1

        matches_played[
            team_b
        ] += 1


    all_teams = sorted(
        set(
            matches[
                "team_a"
            ]
        )
        |
        set(
            matches[
                "team_b"
            ]
        )
    )


    profile_rows = []


    for team in all_teams:

        recent_points = list(
            form_points[
                team
            ]
        )

        recent_goals_for = list(
            goals_for[
                team
            ]
        )

        recent_goals_against = list(
            goals_against[
                team
            ]
        )


        profile_rows.append(
            {
                "team": team,

                "elo":
                    elo[
                        team
                    ],

                "form_points":
                    np.mean(
                        recent_points
                    )
                    if recent_points
                    else 1.0,

                "avg_goals_for":
                    np.mean(
                        recent_goals_for
                    )
                    if recent_goals_for
                    else 1.0,

                "avg_goals_against":
                    np.mean(
                        recent_goals_against
                    )
                    if recent_goals_against
                    else 1.0,

                "win_rate":
                    np.mean(
                        np.array(
                            recent_points
                        )
                        == 3
                    )
                    if recent_points
                    else 0.33,

                "recent_games":
                    len(
                        recent_points
                    ),

                "matches_played":
                    matches_played[
                        team
                    ]
            }
        )


    return (
        pd.DataFrame(
            profile_rows
        )
        .sort_values(
            "elo",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


team_profiles = calculate_current_team_profiles(
    womens_matches
)


projected_2027_teams = (
    team_profiles
    .head(32)
    .copy()
)


print(
    "Projected 2027 field:"
)

display(
    projected_2027_teams
)
```

```python
profile_lookup = (
    team_profiles
    .set_index(
        "team"
    )
    .to_dict(
        orient="index"
    )
)


def create_future_features(
    team_a,
    team_b,
    is_knockout=0
):

    profile_a = profile_lookup[
        team_a
    ]

    profile_b = profile_lookup[
        team_b
    ]


    return pd.DataFrame(
        [
            {
                "elo_diff":
                    profile_a[
                        "elo"
                    ]
                    -
                    profile_b[
                        "elo"
                    ],

                "form_diff":
                    profile_a[
                        "form_points"
                    ]
                    -
                    profile_b[
                        "form_points"
                    ],

                "goals_for_diff":
                    profile_a[
                        "avg_goals_for"
                    ]
                    -
                    profile_b[
                        "avg_goals_for"
                    ],

                "goals_against_diff":
                    profile_a[
                        "avg_goals_against"
                    ]
                    -
                    profile_b[
                        "avg_goals_against"
                    ],

                "win_rate_diff":
                    profile_a[
                        "win_rate"
                    ]
                    -
                    profile_b[
                        "win_rate"
                    ],

                "avg_goals_for_a":
                    profile_a[
                        "avg_goals_for"
                    ],

                "avg_goals_for_b":
                    profile_b[
                        "avg_goals_for"
                    ],

                "avg_goals_against_a":
                    profile_a[
                        "avg_goals_against"
                    ],

                "avg_goals_against_b":
                    profile_b[
                        "avg_goals_against"
                    ],

                "recent_games_a":
                    profile_a[
                        "recent_games"
                    ],

                "recent_games_b":
                    profile_b[
                        "recent_games"
                    ],

                "is_knockout":
                    is_knockout
            }
        ]
    )[
        MODEL_FEATURES
    ]


def predict_2027_match(
    team_a,
    team_b,
    is_knockout=0
):

    features = create_future_features(
        team_a,
        team_b,
        is_knockout
    )


    model_probability_list = []


    for model in [
        logistic_final,
        random_forest_final,
        xgboost_final
    ]:

        probabilities = np.asarray(
            model.predict_proba(
                features
            )[0],
            dtype=np.float64
        )

        probabilities = (
            probabilities
            /
            probabilities.sum()
        )

        model_probability_list.append(
            probabilities
        )


    poisson_features = features[
        POISSON_FEATURES
    ]


    lambda_a = np.clip(
        poisson_team_a_final.predict(
            poisson_features
        )[0],
        0.05,
        8
    )

    lambda_b = np.clip(
        poisson_team_b_final.predict(
            poisson_features
        )[0],
        0.05,
        8
    )


    poisson_probability = (
        calculate_poisson_result_probabilities(
            lambda_a,
            lambda_b
        )
    )


    model_probability_list.append(
        poisson_probability
    )


    ensemble_probability = np.mean(
        model_probability_list,
        axis=0
    )


    ensemble_probability = (
        ensemble_probability
        /
        ensemble_probability.sum()
    )


    return ensemble_probability
```

```python
SIMULATIONS = 1000

rng = np.random.default_rng(
    RANDOM_STATE
)


candidate_teams = (
    projected_2027_teams[
        "team"
    ]
    .tolist()
)


championship_counts = {
    team: 0
    for team in candidate_teams
}


def simulate_knockout_match(
    team_a,
    team_b
):

    probabilities = predict_2027_match(
        team_a,
        team_b,
        is_knockout=1
    )


    # draw probability is split between the two team to represent extra time and penalties
    advance_probability_a = (
        probabilities[
            2
        ]
        +
        0.5
        *
        probabilities[
            1
        ]
    )


    if rng.random() < advance_probability_a:

        return team_a

    return team_b


for simulation_number in range(
    SIMULATIONS
):

    # use Elo order as seeding
    seeded_teams = candidate_teams.copy()


    # randomize teams within four seeding pots
    pots = [
        seeded_teams[
            index:index + 8
        ]
        for index in range(
            0,
            32,
            8
        )
    ]


    for pot in pots:

        rng.shuffle(
            pot
        )


    # each group receives one team from each pot
    groups = [
        [
            pots[
                pot_number
            ][
                group_number
            ]
            for pot_number in range(
                4
            )
        ]
        for group_number in range(
            8
        )
    ]


    knockout_teams = []


    for group in groups:

        group_points = {
            team: 0
            for team in group
        }

        group_goal_score = {
            team: 0.0
            for team in group
        }


        for first_index in range(
            len(
                group
            )
        ):

            for second_index in range(
                first_index + 1,
                len(
                    group
                )
            ):

                team_a = group[
                    first_index
                ]

                team_b = group[
                    second_index
                ]


                probabilities = predict_2027_match(
                    team_a,
                    team_b,
                    is_knockout=0
                )


                result = rng.choice(
                    [
                        0,
                        1,
                        2
                    ],
                    p=probabilities
                )


                group_goal_score[
                    team_a
                ] += probabilities[
                    2
                ] - probabilities[
                    0
                ]

                group_goal_score[
                    team_b
                ] += probabilities[
                    0
                ] - probabilities[
                    2
                ]


                if result == 2:

                    group_points[
                        team_a
                    ] += 3

                elif result == 0:

                    group_points[
                        team_b
                    ] += 3

                else:

                    group_points[
                        team_a
                    ] += 1

                    group_points[
                        team_b
                    ] += 1


        group_ranking = sorted(
            group,
            key=lambda team: (
                group_points[
                    team
                ],
                group_goal_score[
                    team
                ],
                profile_lookup[
                    team
                ][
                    "elo"
                ]
            ),
            reverse=True
        )


        knockout_teams.extend(
            group_ranking[
                :2
            ]
        )


    # round of 16 through the final
    while len(
        knockout_teams
    ) > 1:

        next_round = []


        for index in range(
            0,
            len(
                knockout_teams
            ),
            2
        ):

            winner = simulate_knockout_match(
                knockout_teams[
                    index
                ],
                knockout_teams[
                    index + 1
                ]
            )

            next_round.append(
                winner
            )


        knockout_teams = next_round


    champion = knockout_teams[
        0
    ]

    championship_counts[
        champion
    ] += 1


championship_probabilities = (
    pd.DataFrame(
        {
            "Team":
                list(
                    championship_counts.keys()
                ),

            "Championships":
                list(
                    championship_counts.values()
                )
        }
    )
)


championship_probabilities[
    "Win Probability"
] = (
    championship_probabilities[
        "Championships"
    ]
    /
    SIMULATIONS
)


championship_probabilities = (
    championship_probabilities
    .sort_values(
        "Win Probability",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


print("=" * 70)
print("PREDICTED 2027 WOMEN'S WORLD CUP WINNER")
print("=" * 70)

print(
    "\nPredicted winner:",
    championship_probabilities.loc[
        0,
        "Team"
    ]
)

print(
    "Estimated championship probability:",
    f"{championship_probabilities.loc[0, 'Win Probability']:.2%}"
)


display(
    championship_probabilities.head(
        15
    )
)
```
